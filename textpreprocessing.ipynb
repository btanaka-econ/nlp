```python
import nltk
import string
import re
import inflect
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.stem.porter import PorterStemmer
import csv
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
def text_lowercase(text):
    return text.lower()
def remove_numbers(text):
    return re.sub(r'\d+', '', text)
def remove_punctuation(text):
    translator = str.maketrans('', '', string.punctuation)
    return text.translate(translator)
def remove_whitespace(text):
    return " ".join(text.split())
def remove_stopwords(text):
    stop_words = set(stopwords.words("english"))
    word_tokens = word_tokenize(text)
    filtered_text = [word for word in word_tokens if word.lower()
                     not in stop_words]
    return filtered_text
stemmer = PorterStemmer()
def stem_words(text):
    word_tokens = word_tokenize(text)
    stems = [stemmer.stem(word) for word in word_tokens]
    return stems
            
```

```python
url_filename = 'fomc_url.csv'
with open(url_filename, 'r', encoding='utf-8') as f:
    file = csv.reader(f, delimiter= ',')
    data_list = list(file)
    for i in range (1, 2):
        text_filename = f"{data_list[i][2]}_{data_list[i][3]}.txt"
        with open(text_filename, 'r', encoding='utf-8') as f:
            FOMC_statement = f.read()
            lc = text_lowercase(FOMC_statement)
            #print(lc)
            rmn = remove_numbers(lc)
            #print(rmn)
            rmp = remove_punctuation(rmn)
            #print(rmp)
            rmw = remove_whitespace(rmp)
            #print(rmw)
            rms = remove_stopwords(rmw)
            #print(rms)
            stm = stem_words(rms)
            print(stm)

```
