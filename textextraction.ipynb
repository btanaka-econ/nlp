```python
with open(filename, 'r') as f:
    file = csv.reader(f, delimiter= ',')
    data_list = list(file)
    for i in range(61,189): #61~ from 2006
        url = data_list[i][1]
        # Use "url" as the link to web scrape
        req = Request(url, headers={'User-Agent': 'Mozilla/5.0'})
        page = urlopen(req)
        html = page.read().decode("utf-8")
        soup = BeautifulSoup(html, "html.parser")
        div_list = soup.find_all('div', class_='col-xs-12 col-sm-8 col-md-8')
        for div in div_list:
            all_p_tags = div.find_all('p')
            output_filename = f"{data_list[i][2]}_{data_list[i][3]}.txt"
            with open(output_filename, 'w', encoding='utf-8') as outfile:
                outfile.write(f"Source URL: {url}\n")
                for p_tag in all_p_tags:
                    outfile.write(p_tag.text)
            
```
