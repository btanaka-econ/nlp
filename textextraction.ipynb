```python
from bs4 import BeautifulSoup
import csv
import requests
import sys
from urllib.request import urlopen
```

```python
filename = 'fomc_url.csv'
# Checking the head of csv file
with open(filename, 'r') as f:
    file = csv.reader(f, delimiter=',')
    for i, row in enumerate(file):
        if i < 5:
            print(row)
        else:
            break
```

```python
url = None

with open(filename, 'r') as f:
    file = csv.reader(f, delimiter= ',')
    data_list = list(file)
    url = data_list[1][1]
    # Use "url" as the link to web scrape
    page = urlopen(url)
    html = page.read().decode("utf-8")
    soup = BeautifulSoup(html, "html.parser")
    try:
        response = requests.get(url)
        response.raise_for_status()  # This will raise an error for bad responses (4xx or 5xx)
    
        # 2. Parse the HTML using BeautifulSoup
        soup = BeautifulSoup(response.text, 'html.parser')
    
        # 3. Find all <p> (paragraph) tags
        # This searches the entire document and returns a list of all <p> tags found
        paragraphs = soup.find_all('p')
    
        print(f"Found {len(paragraphs)} paragraph tags. Scraping text...\n")
    
        # 4. Loop through the list and print the text from each one
        for p in paragraphs:
        # .text strips all HTML tags and gives you just the plain text
            print(p.text.strip())
            print("---") # Add a separator to make the output clearer

    except requests.exceptions.RequestException as e:
        print(f"Error: Could not retrieve the webpage. {e}")

```
